# Instrumental Variable (IV)

## Idea

- In causal inference, we wish to isolate the effect of some treatment $D$ on outcome $Y$ using the reduced form $Y = \alpha D + \beta X + \epsilon$

- If $D$ is correlated with the residual $\epsilon$, our treatment effect estimate $\alpha$ is biased. 

- In the IV setting, we try to find some factor $Z$ that is correlated with the treatment $D$ but not with the residuals $\epsilon$. Using this, we "isolate" a component of $D$ that is uncorrelated with $\epsilon$, which then gives us an unbiased coefficient $\alpha$

## Theory of IV

- For an IV to be valid

### Requirements of an IV

- From the examples above, we can kind of gauge what makes a good IV
    1. Relevance: The IV must correlate with the treatment 
        - So for example, birth quarter must predict schooling, which in turn predicts earnings
    2. Exogeneity: The IV must not be correlated with the error term $\epsilon$ in the original equation $Y = \alpha D + \beta X + \epsilon$
        - That is, the IV must be randomly assigned w.r.t the unobserved determinants of $Y$
        - Quarter of birth is not correlated with parental income, upbringing etc
    3. Exclusion restriction: The IV must NOT correlate with the outcome
        - Birth quarter does not predict lifetime earnings besides going through the schooling channel

## Applications of IVs in Research

- Does Compulsory School Attendance Affect Schooling and Earnings - Angrist and Krueger, 1991
    - Wanted to see if additional schooling affects labour market outcomes
    - If we simply do $\text{Wages} \sim \text{Schooling}$, you will end up with biased coefficients
        - Because the decision to undertake more schooling is correlated with things like ability, motivation, and family background, which your model doesn't capture well
    - Since these exogenous effects are hard to capture, the authors make use of a hard policy cutoff to generate an instrumental variable (IV) instead
        - In America, everyone who is born within a calendar year starts school at the same time. That is, whether you are born on 1st Jan or 31st Dec, you start school in the same year
        - However, if you wish to withdraw from school, you need to be 16. For those born in Jan, this happens at the start of the year. For those born in Dec, this happens at the end of the year
        - Therefore, the later you are born in the year, the more expected years of education you accrue 
    - Why this is a good IV?
        - The month of birth is not related to things like health, ability, motivation, family wealth, etc
        - Therefore, the additional education received by people who are born in Dec is effectively randomly selected
    - Thus, instead of $\text{Wages} \sim \text{Schooling}$ 
        - We can use $\hat{\text{Schooling}} = \pi_1 \text{1st Quarter Born} + \pi_2 \text{2nd Quarter Born} + ...$
        - Then take  $\text{Wages} \sim \hat{\text{Schooling}}$ and interpret the coefficient on Birth Quarter as the result of additional education on Wages  

- Lifetime Earnings and the Vietnam Era Draft Lottery - Joshua D. Angrist, 1990
    - Similar idea, wanted ot see what the effect of serving in military is on lifetime earnings
    - But if you use regular military service personnel, there is obvious selection effect. People who choose to enrol in military may have different ability, motivation, and upbringing from those who don't
    - So this paper uses the Vietnam War lottery as an IV
        - People are randomly assigned numbers from 1 to 365. Every year, the military starts drafting people in order of their number from low to high, until they have enough people
        - So the people who serve vs people who don't are effectively randomly selected. This is an IV, and not an RCT, because low values don't automatically mean you serve
    - Why this is a good IV?
        - Lottery is by definition random, so you have randomisation of people who get draft priority
        - So military service is now randomised
    - Thus, instead of $\text{Lifetime earnings} \sim \text{Military}$
        - Take $\hat{\text{Military}} \sim \text{Lottery value}$
        - Then find $\text{Lifetime earnings} \sim \hat{\text{Military}}$
